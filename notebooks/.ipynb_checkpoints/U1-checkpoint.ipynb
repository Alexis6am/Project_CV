## Preparación del entorno y del dataset

Importamos las librerías: OpenCV (`cv`) para procesar las imágenes, NumPy para tratarlas como matrices, Matplotlib para mostrarlas y pandas para organizar la información del dataset en tablas.

```python
import subprocess  

from collections import Counter
from pathlib import Path
import cv2 as cv
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np
```

Fijamos la semilla aleatoria para que los resultados se puedan repetir y definimos la ruta del dataset. Las imágenes están en `data/raw/`, con una carpeta por producto (`producto_antoni`, `producto_cafe`, ...).

```python
np.random.seed(42)

DATOS = Path("../data/raw") #usar r"path" si es que necesitas poner el path completo, aqui no pero

print("Dataset Listo en: ", DATOS.resolve())
print("OpenCV:", cv.__version__, "| Numpy:", np.__version__ )
```

Definimos las funciones auxiliares que usaremos en todo el cuaderno:

- `leer_rgb(ruta)`: lee una imagen y la pasa de BGR (el orden en que la entrega OpenCV) a RGB (el orden que espera Matplotlib).
- `mostrar(imagenes, titulos, cmap, ancho, filas, columnas)`: muestra varias imágenes en una cuadrícula. Las imágenes de un solo canal se dibujan en gris con rango fijo 0-255 para no alterar su contraste.

Después cargamos una imagen de ejemplo de cada producto.

```python
def leer_rgb(ruta):
    #leer en RGB
    return cv.cvtColor(cv.imread(str(ruta)), cv.COLOR_BGR2RGB)

def mostrar(imagenes, titulos, cmap=None, ancho=2.5, filas=2, columnas=5):
    fig, axes = plt.subplots(filas, columnas, figsize=(ancho*5, ancho*2.4))
    for ax, im, tit in zip(np.atleast_1d(axes).ravel(), imagenes, titulos):
        if im.ndim == 2:
            fijo = im.dtype == np.uint8
            ax.imshow(im, cmap=cmap or "gray", vmin=0 if fijo else None, vmax=255 if fijo else None)
        else:
            ax.imshow(im)
        ax.set_title(tit, fontsize=12)
        ax.axis("off")
    plt.tight_layout()
    plt.show()

        
TRAIN = DATOS 

ANTONI = leer_rgb(TRAIN / "producto_antoni/antoni124.jpg")
CAFE = leer_rgb(TRAIN / "producto_cafe/cafe1.jpg")
COCACOLA = leer_rgb(TRAIN / "producto_cocacola/cocacola1.jpg")
COLGATE = leer_rgb(TRAIN / "producto_colgate/colgate1.jpg")
KITKAT = leer_rgb(TRAIN / "producto_kitkat/kitkat1.jpg")
MANZANA = leer_rgb(TRAIN / "producto_manzana/manzana29.jpg")
REAL = leer_rgb(TRAIN / "producto_real/real1.jpg")
SPLENDOR = leer_rgb(TRAIN / "producto_splendor/splendor1.jpg")
TENEGRO = leer_rgb(TRAIN / "producto_tenegro/tenegro1.jpg")
ZEUS = leer_rgb(TRAIN / "producto_zeus/zeus12.jpg")

mostrar(
    [ANTONI, CAFE, COCACOLA, COLGATE, KITKAT, MANZANA, REAL, SPLENDOR, TENEGRO, ZEUS],
    ["Antoni", "Café", "Coca Cola - Lata", "Colgate", "KitKat", "Manzana", "Real", "Splendor", "Té Negro", "Zeus"], 
    None, 2.5,2,5)
```

# Unidad 1: Formación y adquisición de imágenes

En esta unidad vemos cómo se forma y se representa una imagen digital: primero exploramos el dataset, luego comparamos el ojo humano con la cámara, después analizamos la imagen como una matriz de números y terminamos con los espacios de color.

## 1. Exploración del dataset

### 1.1 Construcción del DataFrame

Recorremos cada carpeta de `data/raw/` y guardamos en una tabla la clase (el nombre de la carpeta) y la ruta de cada imagen `.jpg`. La columna `clase_id` asigna un número a cada clase, que es como la entenderá un modelo.

```python
RAIZ = DATOS
FORMATOS_FOTOS = ['.jpg']

lista = [] 
for carpeta in sorted(RAIZ.iterdir()):
    if not carpeta.is_dir():
        continue
    for img in sorted(carpeta.iterdir()):
        if img.suffix.lower() in FORMATOS_FOTOS:
            lista.append({"clase": carpeta.name, "ruta": str(img)})

df = pd.DataFrame(lista) 
df["clase_id"] = df["clase"].astype("category").cat.codes 
```

```python
df.head()
```

### 1.2 División en entrenamiento y prueba

Separamos el dataset en 80 % para entrenamiento y 20 % para prueba. Con `stratify` cada clase conserva la misma proporción en ambos conjuntos, y con `random_state` la división siempre sale igual. Luego unimos las dos partes en un solo DataFrame con una columna `split` que indica a cuál pertenece cada imagen.

```python
from sklearn.model_selection import train_test_split 

train, test = train_test_split(df, test_size= 0.2, stratify=df['clase'], random_state=42) 

#Aqui usar .assign() en vez de usar cosas como train['split'] = 'train', se usa
#Ya que assign retorna un nuevo dataframe en vez de cambiar el original 
df = pd.concat([train.assign(split='train'),test.assign(split='test') ],ignore_index=True)
```

Contamos cuántas imágenes tiene cada clase. La línea naranja marca la media.

```python
conteo = df["clase"].value_counts().sort_values()
print(conteo)
plt.figure(figsize=(8, 10))
plt.barh(conteo.index, conteo.values, color="#42A5F5")
plt.axvline(conteo.mean(), color="orange", ls="--", label=f"media = {conteo.mean():.0f}")
plt.title("Imágenes por categoría (train + test)")
plt.xlabel("Número de imágenes")
plt.legend()
plt.tight_layout()
plt.show()
print(f"Mínimo: {conteo.min()} ({conteo.idxmin()})  |  Máximo: {conteo.max()} ({conteo.idxmax()})")
print(f"Razón de desbalance máx/mín: {conteo.max() / conteo.min():.1f}x")
```

## 2. Sistema de visión humano

La cámara digital imita al ojo: el cristalino funciona como la lente, la pupila como el diafragma y la retina como el sensor. En la retina hay dos tipos de células: los **conos**, que perciben el color, y los **bastones**, que solo perciben la intensidad de la luz. Simulamos tres de estas ideas con nuestras imágenes.

### 2.1 Conos y bastones

Pasar la imagen a escala de grises equivale a quedarnos solo con lo que ven los bastones. Para simular la visión nocturna multiplicamos esa imagen por 0,35.

```python
# Los bastones solo perciben intensidad: equivale a quedarnos con la luminancia de la imagen
bastones = cv.cvtColor(ANTONI, cv.COLOR_RGB2GRAY)
# Simulamos visión nocturna: con poca luz solo trabajan los bastones y todo se ve más oscuro
nocturna = (bastones * 0.35).astype(np.uint8)
mostrar([ANTONI, bastones, nocturna],
        ["Conos (color, buena luz)", "Bastones (solo intensidad)", "Visión nocturna simulada"],None, 2.5, 1, 3)
```

> **Observación.** Sin color se siguen reconociendo las botellas, las etiquetas y el estante: las formas se conservan. Por eso muchos algoritmos trabajan en gris, con un tercio de los datos. Pero se pierde lo que distingue a los vinos entre sí (el color del líquido y de las etiquetas), así que, si el color es lo que separa las clases, no conviene descartarlo.

### 2.2 La fóvea: alta resolución solo en el centro

El ojo solo ve con alta resolución en el centro (la fóvea); la periferia se ve borrosa. Lo simulamos mezclando la imagen original con una versión desenfocada: cuanto más lejos del centro está el píxel, más pesa la versión borrosa. Como nuestras fotos son muy grandes, usamos un desenfoque fuerte (`sigmaX=21`) para que el efecto se note.

```python
h, w = MANZANA.shape[:2]
yy, xx = np.mgrid[:h, :w]
distancia = np.sqrt((yy - h / 2) ** 2 + (xx - w / 2) ** 2) / (0.5 * np.hypot(h, w))
peso = np.clip(distancia * 1.6, 0, 1)[..., None]  # 0 en el centro, 1 en la periferia

periferia = cv.GaussianBlur(MANZANA, (0, 0), sigmaX=21)
foveada = (MANZANA * (1 - peso) + periferia * peso).astype(np.uint8)
mostrar([MANZANA, foveada], ["Cámara: resolución uniforme", "Ojo: fóvea nítida, periferia borrosa"], None, 2.5, 1,2)
```

> **Observación.** Las manzanas que están en la mano quedan nítidas y el resto se desenfoca. La cámara, en cambio, captura toda la escena con la misma resolución, lo que obliga a procesar todos los píxeles por igual.

### 2.3 La corteza V1 detecta bordes y orientaciones

Las primeras neuronas de la corteza visual (V1) responden a bordes con una orientación determinada. Lo simulamos con el filtro Sobel, que detecta cambios de intensidad en una dirección. Escalamos el resultado al percentil 99 para que unos pocos bordes muy fuertes no oscurezcan el resto.

```python
gris = cv.cvtColor(SPLENDOR, cv.COLOR_RGB2GRAY)
bordes_verticales = np.abs(cv.Sobel(gris, cv.CV_64F, 1, 0, ksize=3))
bordes_horizontales = np.abs(cv.Sobel(gris, cv.CV_64F, 0, 1, ksize=3))
# Escalamos al percentil 99 para que unos pocos bordes muy fuertes no oscurezcan el resto
bordes_verticales = np.clip(bordes_verticales / np.percentile(bordes_verticales, 99), 0, 1)
bordes_horizontales = np.clip(bordes_horizontales / np.percentile(bordes_horizontales, 99), 0, 1)
mostrar([gris, bordes_verticales, bordes_horizontales],
        ["Entrada (retina)", "Neuronas sensibles a bordes verticales", "Neuronas sensibles a bordes horizontales"],
        "gray", 2.5, 1, 3)
```

> **Observación.** Un mapa resalta los bordes verticales (los lados de las botellas y del empaque) y el otro los horizontales (el estante y el texto de la etiqueta de precio). Las primeras capas de una red neuronal convolucional aprenden filtros muy parecidos a estos.

## 3. Sistema de visión artificial: el sensor y el filtro de Bayer

Cada píxel del sensor de una cámara solo mide cuánta luz le llega, no su color. Para capturar color se coloca encima un mosaico de filtros llamado **patrón de Bayer** (RGGB), con el doble de filtros verdes porque el ojo es más sensible al verde. Cada píxel mide un solo canal y los otros dos se calculan a partir de sus vecinos; a este proceso se le llama **demosaicing**. Lo simulamos con un recorte de 80×80 píxeles de la imagen del Té Negro.

```python
recorte = TENEGRO[2000:2080, 600:680]  # recorte de 80x80 para ver el mosaico
h, w = recorte.shape[:2]

# Lo que mide el sensor: una sola matriz 2D, cada píxel con un único canal según el patrón RGGB
crudo = np.zeros((h, w), np.uint8)
crudo[0::2, 0::2] = recorte[0::2, 0::2, 0]  # R
crudo[0::2, 1::2] = recorte[0::2, 1::2, 1]  # G
crudo[1::2, 0::2] = recorte[1::2, 0::2, 1]  # G
crudo[1::2, 1::2] = recorte[1::2, 1::2, 2]  # B

# Visualización del mosaico: cada píxel pintado del color de su filtro
mosaico = np.zeros_like(recorte)
mosaico[0::2, 0::2, 0] = crudo[0::2, 0::2]
mosaico[0::2, 1::2, 1] = crudo[0::2, 1::2]
mosaico[1::2, 0::2, 1] = crudo[1::2, 0::2]
mosaico[1::2, 1::2, 2] = crudo[1::2, 1::2]

reconstruida = cv.cvtColor(crudo, cv.COLOR_BayerRGGB2RGB)  # demosaicing
error = np.abs(reconstruida.astype(int) - recorte.astype(int)).mean()

mostrar([recorte, mosaico, crudo, reconstruida],
        ["Escena real", "Mosaico de Bayer (RGGB)", "Dato crudo del sensor (1 canal)",
         f"Demosaicing (error medio {error:.1f})"],"gray", 2.5, 1, 4)
print("Forma del dato crudo:", crudo.shape, "| Forma reconstruida:", reconstruida.shape)
```

> **Observación.** El sensor entrega una sola matriz 2D, y aun así la reconstrucción a color se parece mucho a la escena original: el error medio es de unos pocos niveles sobre 255. Las diferencias aparecen en los bordes marcados, donde se ven pequeños artefactos de color. Todas las fotos del dataset pasaron por este proceso dentro del celular.

## 4. Digitalización: muestreo, cuantificación y profundidad de color

### 4.1 La imagen es una matriz

Una imagen digital es una matriz de números. A color tiene forma `(alto, ancho, 3)` y en gris `(alto, ancho)`. Cada valor es un entero entre 0 y 255 (tipo `uint8`). Mostramos un parche de 15×15 píxeles para ver los números directamente.

```python
print("Forma:", MANZANA.shape, "→ (alto, ancho, canales)")
print("Tipo de dato:", MANZANA.dtype, "→ enteros sin signo de 8 bits, rango",
      MANZANA.min(), "a", MANZANA.max())
print("Número de valores:", MANZANA.size, "=", " × ".join(map(str, MANZANA.shape)))

gris = cv.cvtColor(MANZANA, cv.COLOR_RGB2GRAY)
fila, col = 1800, 1000
parche = gris[fila:fila + 15, col:col + 15]
print(f"\nPíxel RGB en (fila={fila}, col={col}):", MANZANA[fila, col])
print("\nParche 15x15 en escala de grises (una tabla de datos de 15 filas y 15 columnas):")
print(parche)

fig, ejes = plt.subplots(1, 2, figsize=(9, 4.5))
ejes[0].imshow(gris, cmap="gray")
ejes[0].add_patch(plt.Rectangle((col, fila), 15, 15, edgecolor="red", fill=False, lw=1))
ejes[0].set_title("Imagen en gris y ubicación del parche")
ejes[1].imshow(parche, cmap="gray", vmin=0, vmax=255)
for (i, j), v in np.ndenumerate(parche):
    ejes[1].text(j, i, v, ha="center", va="center", color="red", fontsize=8)
ejes[1].set_title("Parche 15x15: cada celda es un número")
for e in ejes:
    e.axis("off")
plt.tight_layout()
plt.show()
```

> **Observación.** Cada celda del parche es un valor de intensidad, y los píxeles vecinos tienen valores parecidos. A diferencia de una tabla de datos normal, aquí la posición importa. Todas las operaciones que siguen son, en el fondo, cálculos sobre esta matriz.

### 4.2 Muestreo: resolución espacial

El **muestreo** decide cuántos píxeles tiene la imagen. Reducimos la Manzana a varias resoluciones y la volvemos a ampliar sin interpolar, para ver cada píxel como un bloque.

```python
resoluciones = [348, 64, 32, 16]
versiones = []
for r in resoluciones:
    pequena = cv.resize(MANZANA, (r, r), interpolation=cv.INTER_AREA)
    # Volvemos sin interpolar, para ver los píxeles como bloques
    versiones.append(cv.resize(pequena, (338, 700), interpolation=cv.INTER_NEAREST))
mostrar(versiones, [f"{r}x{r} píxeles" for r in resoluciones], None, 3.5, 1, 4)
```

> **Observación.** Al bajar la resolución se pierden los detalles finos, como la textura de la cáscara o los dedos. A 64×64 las manzanas todavía se distinguen, a 32×32 apenas, y a 16×16 solo quedan manchas de color. Elegir la resolución de entrada de un modelo es un equilibrio entre detalle y costo de cómputo.

### 4.3 Cuantificación y profundidad de color

La **cuantificación** decide cuántos niveles de intensidad puede tomar cada píxel. Con 8 bits por canal hay 256 niveles; con menos bits, cada valor se lleva al nivel más cercano. Lo probamos con la lata de Coca-Cola.

```python
def cuantificar(img, bits):
    niveles = 2 ** bits
    paso = 256 // niveles
    # Cada valor se lleva al centro de su intervalo: 256 niveles → 'niveles' niveles
    return (img // paso * paso + paso // 2).astype(np.uint8)


filas = []
for bits in [8, 4, 2, 1]:
    q = cuantificar(COCACOLA, bits)
    colores = len(np.unique(q.reshape(-1, 3), axis=0))
    filas.append({"bits por canal": bits, "niveles por canal": 2 ** bits,
                  "colores posibles": (2 ** bits) ** 3, "colores usados": colores,
                  "tamaño en memoria (KB)": COCACOLA.shape[0] * COCACOLA.shape[1] * 3 * bits / 8 / 1024})
mostrar([cuantificar(COCACOLA, b) for b in [8, 4, 2, 1]],
        ["8 bits/canal", "4 bits/canal", "2 bits/canal", "1 bit/canal"], None, 3.5, 1,4)
pd.DataFrame(filas).round(1)
```

> **Observación.** Con 4 bits casi no se nota la diferencia, pero con 2 bits aparecen bandas de color en zonas que deberían ser suaves, como el fondo y el estante; a esto se le llama **posterización**. Con 1 bit por canal solo quedan 8 colores posibles. Menos bits ocupan menos memoria, pero se pierde precisión.

### 4.4 Formatos de archivo: compresión con y sin pérdida

Guardamos la imagen en memoria en distintos formatos y medimos su tamaño y el error al volver a leerla. BMP y PNG comprimen **sin pérdida**; JPEG comprime **con pérdida** y permite ajustar la calidad.

```python
bgr = cv.cvtColor(COCACOLA, cv.COLOR_RGB2BGR)
filas, decodificadas = [], []
for nombre, ext, params in [("BMP", ".bmp", []), ("PNG", ".png", []),
                            ("JPEG calidad 90", ".jpg", [cv.IMWRITE_JPEG_QUALITY, 90]),
                            ("JPEG calidad 30", ".jpg", [cv.IMWRITE_JPEG_QUALITY, 30]),
                            ("JPEG calidad 5", ".jpg", [cv.IMWRITE_JPEG_QUALITY, 5])]:
    ok, buffer = cv.imencode(ext, bgr, params)
    recuperada = cv.cvtColor(cv.imdecode(buffer, cv.IMREAD_COLOR), cv.COLOR_BGR2RGB)
    decodificadas.append((nombre, recuperada))
    filas.append({"formato": nombre, "tamaño (KB)": buffer.size / 1024,
                  "error medio por píxel": np.abs(recuperada.astype(int) - COCACOLA.astype(int)).mean()})

zoom = (slice(2000, 2200), slice(1500, 1800))
mostrar([COCACOLA[zoom], decodificadas[3][1][zoom], decodificadas[4][1][zoom]],
        ["Original (zoom)", "JPEG calidad 30 (zoom)", "JPEG calidad 5 (zoom)"],None, 3.5, 1,3)
pd.DataFrame(filas).round(2)
```

> **Observación.** BMP y PNG recuperan la imagen exacta (error 0), y PNG ocupa bastante menos que BMP. JPEG ocupa mucho menos, pero el error crece al bajar la calidad: con calidad 5, el zoom muestra bloques y bordes con colores falsos. Conviene guardar las fotos originales y no volver a guardarlas en JPEG varias veces, porque la pérdida se acumula.

## 5. Espacios de color

### 5.1 Escala de grises

La conversión a gris no es un promedio simple: le da más peso al verde porque el ojo es más sensible a ese color.

$$Gris = 0.299\,R + 0.587\,G + 0.114\,B$$

Comparamos la fórmula escrita a mano con OpenCV y con el promedio simple.

```python
R, G, B = [MANZANA[..., i].astype(float) for i in range(3)]
gris_manual = (0.299 * R + 0.587 * G + 0.114 * B).round().astype(np.uint8)
gris_opencv = cv.cvtColor(MANZANA, cv.COLOR_RGB2GRAY)
gris_promedio = ((R + G + B) / 3).round().astype(np.uint8)

print("Diferencia máxima fórmula manual vs OpenCV:",
      np.abs(gris_manual.astype(int) - gris_opencv.astype(int)).max())
print("Diferencia media promedio simple vs ponderado:",
      np.abs(gris_promedio.astype(int) - gris_opencv.astype(int)).mean().round(1))
mostrar([gris_opencv, gris_promedio], ["Gris ponderado (0.299R+0.587G+0.114B)", "Promedio simple (R+G+B)/3"],
        "gray",3.5, 1,2)
```

> **Observación.** La fórmula manual coincide con OpenCV (como mucho 1 nivel de diferencia, por redondeo). El promedio simple da un resultado parecido pero no igual, porque reparte el peso por igual entre los tres canales.

### 5.2 Blanco y negro (imagen binaria)

Una imagen binaria solo tiene dos valores: 0 (negro) y 255 (blanco). Se obtiene con un **umbral**: los píxeles que lo superan pasan a blanco y el resto a negro. Probamos un umbral fijo de 127 y el método de **Otsu**, que calcula el umbral automáticamente a partir del histograma.

```python
gris = cv.cvtColor(MANZANA, cv.COLOR_RGB2GRAY)
_, binaria_127 = cv.threshold(gris, 127, 255, cv.THRESH_BINARY)
umbral_otsu, binaria_otsu = cv.threshold(gris, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)
mostrar([gris, binaria_127, binaria_otsu],
        ["Escala de grises", "Umbral fijo = 127", f"Umbral automático (Otsu) = {umbral_otsu:.0f}"], "gray", 2.4, 1,3)
print("Valores únicos en la imagen binaria:", np.unique(binaria_127))
```

> **Observación.** Los dos umbrales dan resultados casi iguales y ninguno aísla la manzana: las zonas claras de muchas manzanas quedan blancas y las sombras quedan negras. Un único umbral para toda la imagen no funciona bien cuando la iluminación varía. En la Unidad 2 (sección 1.3) se repite esta comparación con otros productos y se añade el umbral adaptativo.

### 5.3 RGB: canales e histogramas

En RGB cada píxel tiene tres valores: rojo, verde y azul. Mostramos cada canal por separado, a color y como intensidad, junto con el histograma de cada uno.

```python
fig, ejes = plt.subplots(2, 4, figsize=(16, 7))
ejes[0, 0].imshow(MANZANA)
ejes[0, 0].set_title("Imagen RGB")
colores = ["red", "green", "blue"]
for i, (nombre, color) in enumerate(zip(["R", "G", "B"], colores)):
    canal = np.zeros_like(MANZANA)
    canal[..., i] = MANZANA[..., i]
    ejes[0, i + 1].imshow(canal)
    ejes[0, i + 1].set_title(f"Canal {nombre}")
    ejes[1, i + 1].imshow(MANZANA[..., i], cmap="gray")
    ejes[1, i + 1].set_title(f"Canal {nombre} como intensidad")
    hist = cv.calcHist([MANZANA], [i], None, [256], [0, 256]).ravel()
    ejes[1, 0].plot(hist, color=color, label=nombre)
ejes[1, 0].set_title("Histograma por canal")
ejes[1, 0].legend()
for e in list(ejes[0]) + list(ejes[1, 1:]):
    e.axis("off")
plt.tight_layout()
plt.show()
```

> **Observación.** El canal R es claro en casi todas las manzanas y su histograma está desplazado hacia valores altos, porque el rojo domina la escena. El canal B es el más oscuro. Las manzanas amarillas se ven claras en R y también en G, porque en el modelo aditivo rojo + verde = amarillo.

### 5.4 HSV: matiz, saturación y brillo

HSV describe el color como lo haría una persona:

- **H (matiz):** qué color es. En OpenCV va de 0 a 179, y el rojo está en los dos extremos (≈0 y ≈179).
- **S (saturación):** qué tan puro es el color (0 = gris).
- **V (brillo):** qué tan claro es (0 = negro).

```python
hsv = cv.cvtColor(MANZANA, cv.COLOR_RGB2HSV)
fig, ejes = plt.subplots(1, 4, figsize=(17, 4.2))
ejes[0].imshow(MANZANA)
ejes[0].set_title("RGB")
mapa = ejes[1].imshow(hsv[..., 0], cmap="hsv", vmin=0, vmax=179)
ejes[1].set_title("H: matiz (0-179)")
plt.colorbar(mapa, ax=ejes[1], fraction=0.046)
ejes[2].imshow(hsv[..., 1], cmap="gray")
ejes[2].set_title("S: saturación")
ejes[3].imshow(hsv[..., 2], cmap="gray")
ejes[3].set_title("V: brillo")
for e in ejes:
    e.axis("off")
plt.tight_layout()
plt.show()
```

> **Observación.** En H casi toda la imagen aparece en tonos rojo-naranja, los de las manzanas. La manga y el fondo oscuro muestran matices mezclados porque, cuando la saturación es baja, el matiz no es confiable. S separa lo que tiene color de lo que es gris, y V se parece a la imagen en gris: concentra la información de iluminación.

### 5.5 El problema de la iluminación: RGB frente a HSV

Simulamos menos luz multiplicando la imagen por 0,45 y comparamos el valor del mismo píxel en ambos espacios.

```python
oscura = (MANZANA.astype(float) * 0.45).astype(np.uint8)
fila, col = 140, 120  # un píxel sobre un pimiento rojo
hsv_clara = cv.cvtColor(MANZANA, cv.COLOR_RGB2HSV)
hsv_oscura = cv.cvtColor(oscura, cv.COLOR_RGB2HSV)

pd.DataFrame(
    [MANZANA[fila, col].tolist() + hsv_clara[fila, col].tolist(),
     oscura[fila, col].tolist() + hsv_oscura[fila, col].tolist()],
    index=["Luz normal", "Poca luz"], columns=["R", "G", "B", "H", "S", "V"],
)
```

> **Observación.** En RGB los tres valores cambian al bajar la luz. En HSV solo cambia V, mientras que H y S se mantienen iguales. Por eso HSV es mejor para segmentar por color.

### 5.6 Experimento: segmentar la manzana en RGB y en HSV

Creamos dos máscaras para seleccionar los píxeles rojos: una con reglas sobre R, G y B, y otra con rangos de H, S y V (el rojo se toma en los dos extremos del matiz). Las aplicamos con luz normal y con poca luz, y medimos cuánto se parecen con el **IoU** (intersección sobre unión; 1 = máscaras idénticas).

```python
def mascara_rgb(img):
    R, G, B = [img[..., i].astype(int) for i in range(3)]
    return ((R > 150) & (G < 90) & (B < 90)).astype(np.uint8) * 255


def mascara_hsv(img):
    hsv = cv.cvtColor(img, cv.COLOR_RGB2HSV)
    # El rojo está en ambos extremos del matiz: unimos los dos rangos
    bajo = cv.inRange(hsv, (0, 120, 40), (8, 255, 255))
    alto = cv.inRange(hsv, (170, 120, 40), (179, 255, 255))
    return cv.bitwise_or(bajo, alto)


def iou(a, b):
    a, b = a > 0, b > 0
    return (a & b).sum() / max((a | b).sum(), 1)


resultados = {}
fig, ejes = plt.subplots(2, 3, figsize=(13, 8.5))
for f, (nombre, img) in enumerate([("Luz normal", MANZANA), ("Poca luz", oscura)]):
    m_rgb, m_hsv = mascara_rgb(img), mascara_hsv(img)
    resultados[nombre] = (m_rgb, m_hsv)
    ejes[f, 0].imshow(img)
    ejes[f, 0].set_title(nombre)
    ejes[f, 1].imshow(m_rgb, cmap="gray")
    ejes[f, 1].set_title(f"Máscara RGB ({(m_rgb > 0).mean():.0%} de la imagen)")
    ejes[f, 2].imshow(m_hsv, cmap="gray")
    ejes[f, 2].set_title(f"Máscara HSV ({(m_hsv > 0).mean():.0%} de la imagen)")
for e in ejes.ravel():
    e.axis("off")
plt.tight_layout()
plt.show()

print("Consistencia entre luz normal y poca luz (IoU, 1 = máscaras idénticas):")
print(f"  RGB: {iou(resultados['Luz normal'][0], resultados['Poca luz'][0]):.2f}")
print(f"  HSV: {iou(resultados['Luz normal'][1], resultados['Poca luz'][1]):.2f}")
```

```python
segmentado = cv.bitwise_and(MANZANA, MANZANA, mask=mascara_hsv(MANZANA))
mostrar([MANZANA, segmentado], ["Original", "Solo los píxeles rojos (máscara HSV)"], None, 3, 1,2)
```

> **Observación.** Con poca luz la máscara RGB desaparece por completo, porque exige R > 150 y ya ningún píxel lo cumple. La máscara HSV encuentra casi los mismos píxeles en ambos casos (IoU alto). Al aplicar la máscara HSV sobre la imagen original quedan solo las zonas rojas de las manzanas; las partes amarillas y los brillos se descartan porque su matiz o su saturación caen fuera del rango.

### 5.7 Conversión entre espacios de color

Convertimos de RGB a HSV y de vuelta a RGB para comprobar cuánto se pierde, y resumimos en una tabla los espacios de color vistos.

```python
ida_vuelta = cv.cvtColor(cv.cvtColor(MANZANA, cv.COLOR_RGB2HSV), cv.COLOR_HSV2RGB)
print("Error medio RGB → HSV → RGB:", np.abs(ida_vuelta.astype(int) - MANZANA.astype(int)).mean().round(2))

pd.DataFrame([
    ["Gris", 1, "No aplica", "—", "Filtros, bordes, descriptores clásicos", "cv.COLOR_RGB2GRAY"],
    ["Binaria", 1, "No aplica", "—", "Aislar regiones de interés", "cv.threshold"],
    ["RGB", 3, "No", "Baja", "Captura y visualización", "cv.COLOR_BGR2RGB"],
    ["HSV", 3, "Sí", "Alta", "Segmentación por color", "cv.COLOR_RGB2HSV"],
], columns=["Espacio", "Canales", "Separa color y brillo", "Robustez a iluminación", "Uso típico", "Conversión en OpenCV"])
```

> **Observación.** El error de ida y vuelta es menor a 1 nivel; se debe al redondeo, porque H se guarda en el rango 0-179. Se puede procesar en HSV y volver a RGB para mostrar el resultado sin perder calidad. La Unidad 2 usa esto para ecualizar solo el brillo de una imagen a color.

## 6. El color como característica

Resumimos cada imagen con tres números: su H, S y V medios. Solo usamos los píxeles con saturación mayor a 60, para ignorar los fondos grises o blancos (si una imagen no tiene ninguno, se usan todos). Como el matiz es un ángulo (179 está junto a 0), su promedio se calcula como **media circular**. Calculamos estos valores para 30 imágenes de entrenamiento de cada clase y los graficamos.

```python
def color_medio_hsv(img):
    hsv = cv.cvtColor(cv.resize(img, (128, 128)), cv.COLOR_RGB2HSV).reshape(-1, 3).astype(float)
    hsv_color = hsv[hsv[:, 1] > 60]  # solo píxeles con color (ignora fondo gris/blanco)
    
    # Salvaguarda por si algún empaque es casi todo blanco/negro y no supera S > 60
    if len(hsv_color) == 0:
        hsv_color = hsv
        
    angulo = hsv_color[:, 0] * 2 * np.pi / 180
    h_medio = (np.degrees(np.arctan2(np.sin(angulo).mean(), np.cos(angulo).mean())) % 360) / 2
    return h_medio, hsv_color[:, 1].mean(), hsv_color[:, 2].mean()


# 1. Tus 10 clases (excluimos segmentacion_colgate porque no está en tu tabla de clases)
productos = [
    "producto_antoni",
    "producto_cafe",
    "producto_cocacola",
    "producto_colgate",
    "producto_kitkat",
    "producto_manzana",
    "producto_real",
    "producto_splendor",
    "producto_tenegro",
    "producto_zeus"
]

# 2. Paleta de 10 colores distintos (tab10 de Matplotlib)
colores_graf = plt.cm.tab10.colors

filas = []
for producto in productos:
    # Filtramos usando la columna 'clase' directamente
    rutas = df[(df["split"] == "train") & (df["clase"] == producto)]["ruta"].head(30)
    for r in rutas:
        # Como 'r' ya incluye '..\data\raw\...', la pasamos directo (o usa Path(r))
        h, s, v = color_medio_hsv(leer_rgb(r))
        filas.append({"producto": producto, "H": h, "S": s, "V": v})

caracteristicas = pd.DataFrame(filas)

fig, ejes = plt.subplots(1, 2, figsize=(14, 5.5))
for producto, color in zip(productos, colores_graf):
    sub_df = caracteristicas[caracteristicas["producto"] == producto]
    ejes[0].scatter(sub_df["H"], sub_df["S"], color=color, label=producto, alpha=0.75, edgecolor="k")
    ejes[1].scatter(sub_df["H"], sub_df["V"], color=color, label=producto, alpha=0.75, edgecolor="k")

ejes[0].set(xlabel="H medio (0-179)", ylabel="S media", title="Matiz vs saturación")
ejes[1].set(xlabel="H medio (0-179)", ylabel="V media", title="Matiz vs brillo")
ejes[1].legend(bbox_to_anchor=(1.02, 1), loc="upper left")  # leyenda afuera para no tapar puntos
plt.tight_layout()
plt.show()

caracteristicas.groupby("producto")[["H", "S", "V"]].agg(["mean", "std"]).round(1)
```

> **Observación.** Si los puntos de cada clase forman grupos separados, el color medio basta para distinguirlas; si se solapan, hacen falta otras características (forma, textura) que se verán en las siguientes unidades. Esto ya es una **extracción de características**: cada imagen pasa de millones de píxeles a solo tres números.

# Unidad 2: Operaciones y preprocesamiento de imágenes

El objetivo es aplicar operaciones de procesamiento digital de imágenes para mejorar su calidad y facilitar su análisis posterior. En particular, se trabaja con ecualización de histograma, CLAHE, umbralización, convolución, filtros para reducción de ruido y operaciones morfológicas. Para observar cómo se comportan estas técnicas en distintos productos, se utilizan tres ejemplos del dataset: **Antoni, KitKat y Zeus**.

Las operaciones no cambian la categoría del producto; transforman su representación visual para hacer más visibles ciertos detalles o reducir elementos que pueden dificultar un análisis automático. Los resultados servirían como una etapa inicial antes de extraer características o entrenar un modelo de clasificación.

## 1. Operaciones sobre puntos: histograma y umbral

```python
def histograma(eje, gris, titulo, color="#42A5F5"):
    eje.hist(gris.ravel(), bins=256, range=(0, 256), color=color)
    eje.set_title(titulo)
    eje.set_xlim(0, 255)
    eje.set_yticks([])
```

### 1.1 Normalización de histograma (estiramiento de contraste)

```python
gris = cv.cvtColor(KITKAT, cv.COLOR_RGB2GRAY)
subexpuesta = (gris * 0.35 + 30).astype(np.uint8)  # simulación: todo cae entre 30 y ~120

minimo, maximo = int(subexpuesta.min()), int(subexpuesta.max())
manual = ((subexpuesta.astype(float) - minimo) / (maximo - minimo) * 255).astype(np.uint8)
opencv = cv.normalize(subexpuesta, None, 0, 255, cv.NORM_MINMAX)

print(f"Rango antes: [{minimo}, {maximo}]  |  Rango después: [{manual.min()}, {manual.max()}]")
print("Diferencia máxima fórmula manual vs cv2.normalize:",
      np.abs(manual.astype(int) - opencv.astype(int)).max())

fig, ejes = plt.subplots(2, 2, figsize=(10, 8))
ejes[0, 0].imshow(subexpuesta, cmap="gray", vmin=0, vmax=255)
ejes[0, 0].set_title("Subexpuesta")
ejes[0, 1].imshow(opencv, cmap="gray", vmin=0, vmax=255)
ejes[0, 1].set_title("Normalizada")
histograma(ejes[1, 0], subexpuesta, "Histograma: concentrado en 30-120")
histograma(ejes[1, 1], opencv, "Histograma: estirado a 0-255")
for e in ejes[0]:
    e.axis("off")
plt.tight_layout()
plt.show()
```

### 1.2 Ecualización de histograma y CLAHE

```python
# Seleccionamos tres productos para comparar el efecto de la ecualización.
ejemplos = [("Antoni", ANTONI), ("KitKat", KITKAT), ("Zeus", ZEUS)]

for nombre_producto, imagen in ejemplos:
    # Convertimos la imagen a escala de grises para trabajar con sus intensidades.
    gris_oscura = cv.cvtColor(imagen, cv.COLOR_RGB2GRAY)

    # Calculamos el histograma y la función de distribución acumulada (CDF).
    hist = np.bincount(gris_oscura.ravel(), minlength=256)
    cdf = hist.cumsum()
    cdf_min = cdf[cdf > 0][0]

    # Construimos manualmente la tabla de ecualización y la aplicamos a cada píxel.
    tabla = np.round((cdf - cdf_min) / (gris_oscura.size - cdf_min) * 255).clip(0, 255).astype(np.uint8)
    ecualizada_manual = tabla[gris_oscura]
    ecualizada = cv.equalizeHist(gris_oscura)

    # Comprobamos que la implementación manual coincide con la función de OpenCV.
    print(f"{nombre_producto} - diferencia máxima ecualización manual vs OpenCV:",
          np.abs(ecualizada_manual.astype(int) - ecualizada.astype(int)).max())

    # CLAHE ecualiza por pequeñas regiones y limita el contraste para evitar amplificar demasiado el ruido.
    clahe = cv.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8)).apply(gris_oscura)

    # Comparamos imagen original, ecualización global y CLAHE, junto con sus histogramas y CDF.
    fig, ejes = plt.subplots(3, 3, figsize=(13, 11))
    for c, (img, titulo) in enumerate([
        (gris_oscura, "Original"),
        (ecualizada, "Ecualización global"),
        (clahe, "CLAHE (8x8 bloques)")
    ]):
        ejes[0, c].imshow(img, cmap="gray", vmin=0, vmax=255)
        ejes[0, c].set_title(titulo)
        ejes[0, c].axis("off")

        # El histograma muestra la distribución de intensidades de cada resultado.
        histograma(ejes[1, c], img, "Histograma")
        cdf_img = np.bincount(img.ravel(), minlength=256).cumsum()
        ejes[2, c].plot(cdf_img / cdf_img[-1], color="orange")
        ejes[2, c].set_title("CDF (acumulada)")
        ejes[2, c].set_xlim(0, 255)

    plt.suptitle(f"Ecualización de histograma - {nombre_producto}")
    plt.tight_layout()
    plt.show()

    # Revisamos el brillo medio y la desviación estándar como referencia del contraste.
    for titulo, img in [("Original", gris_oscura), ("Ecualizada", ecualizada), ("CLAHE", clahe)]:
        print(f"{nombre_producto} - {titulo:<11} media = {img.mean():6.1f}   desviación estándar = {img.std():5.1f}")
```

**Ecualizar una imagen a color.** Si ecualizas R, G y B por separado, los colores cambian. Lo
correcto es convertir a HSV (Unidad 1), ecualizar solo el canal **V** (brillo) y volver a RGB (en la sección 5.7 de la Unidad 1 comprobamos que esta ida y vuelta casi no pierde información).

```python
# Comparamos dos formas de ecualizar imágenes a color en tres productos.
ejemplos = [("Antoni", ANTONI), ("KitKat", KITKAT), ("Zeus", ZEUS)]

for nombre_producto, imagen in ejemplos:
    # Convertimos RGB a HSV para separar el tono y la saturación del brillo.
    hsv = cv.cvtColor(imagen, cv.COLOR_RGB2HSV)
    hsv_v = hsv.copy()

    # Aplicamos CLAHE únicamente al canal V (brillo), conservando tono y saturación.
    hsv_v[..., 2] = cv.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8)).apply(hsv[..., 2])
    color_v = cv.cvtColor(hsv_v, cv.COLOR_HSV2RGB)

    # Esta alternativa ecualiza R, G y B por separado; puede alterar los colores originales.
    por_canal = np.dstack([cv.equalizeHist(imagen[..., i]) for i in range(3)])

    # Visualizamos las diferencias entre la imagen original y los dos métodos.
    mostrar([imagen, por_canal, color_v],
            [f"{nombre_producto}: original", "Ecualización R, G y B", "CLAHE solo en V (HSV)"],
            None, 2.5, 1, 3)
```

### 1.3 Binarización por umbral: fijo, Otsu y adaptativo

En la Unidad 1 (sección 5.2) se aplicaron el umbral fijo y Otsu a la Manzana. Aquí se repiten con tres productos y se añade el umbral **adaptativo**, que calcula un umbral distinto para cada píxel a partir de su vecindario.

```python
# Aplicamos tres métodos de umbralización a cada producto.
ejemplos = [("Antoni", ANTONI), ("KitKat", KITKAT), ("Zeus", ZEUS)]

for nombre_producto, imagen in ejemplos:
    # La umbralización se realiza sobre una imagen en escala de grises.
    gris = cv.cvtColor(imagen, cv.COLOR_RGB2GRAY)

    # Umbral fijo: utiliza el mismo valor de intensidad para todos los píxeles.
    _, fijo = cv.threshold(gris, 127, 255, cv.THRESH_BINARY)

    # Otsu calcula automáticamente un umbral global a partir del histograma.
    t_otsu, otsu = cv.threshold(gris, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

    # El umbral adaptativo calcula un valor local usando el vecindario de cada píxel.
    adaptativo = cv.adaptiveThreshold(
        gris, 255, cv.ADAPTIVE_THRESH_GAUSSIAN_C, cv.THRESH_BINARY, 31, 5
    )

    # Mostramos el histograma y el resultado de los tres métodos.
    fig, ejes = plt.subplots(1, 4, figsize=(17, 4.5))
    histograma(ejes[0], gris, f"{nombre_producto}: histograma")
    ejes[0].axvline(127, color="red", ls="--", label="fijo = 127")
    ejes[0].axvline(t_otsu, color="orange", label=f"Otsu = {t_otsu:.0f}")
    ejes[0].legend()

    for eje, img, titulo in zip(
        ejes[1:], [fijo, otsu, adaptativo],
        ["Umbral fijo", "Otsu (global)", "Adaptativo (local)"]
    ):
        eje.imshow(img, cmap="gray")
        eje.set_title(titulo)
        eje.axis("off")

    plt.tight_layout()
    plt.show()
```

## 2. Operaciones sobre grupos de puntos: convolución y filtros

### 2.1 Convolución paso a paso

```python
def convolucion(imagen, kernel):
    k = kernel.shape[0] // 2
    # Relleno (padding) replicando el borde, para que la salida tenga el mismo tamaño
    con_borde = np.pad(imagen.astype(float), k, mode="edge")
    salida = np.zeros(imagen.shape, dtype=float)
    for y in range(imagen.shape[0]):
        for x in range(imagen.shape[1]):
            vecindad = con_borde[y:y + 2 * k + 1, x:x + 2 * k + 1]
            salida[y, x] = (vecindad * kernel).sum()
    return salida


pequena = np.array([[10, 10, 10, 200, 200, 200],
                    [10, 10, 10, 200, 200, 200],
                    [10, 10, 10, 200, 200, 200],
                    [10, 10, 10, 200, 200, 200],
                    [10, 10, 10, 200, 200, 200],
                    [10, 10, 10, 200, 200, 200]], dtype=np.uint8)
media_3x3 = np.ones((3, 3)) / 9
sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]], dtype=float)

print("Imagen 6x6 con un borde vertical:\n", pequena)
print("\nConvolución con kernel de media 3x3:\n", convolucion(pequena, media_3x3).round(0))
print("\nConvolución con Sobel x (detecta cambios horizontales de intensidad):\n", convolucion(pequena, sobel_x))
```

```python
# Definimos varios kernels para comparar sus efectos sobre las imágenes.
kernels = {
    "Identidad": np.array([[0, 0, 0], [0, 1, 0], [0, 0, 0]], dtype=float),
    "Media 5x5": np.ones((5, 5)) / 25,
    "Enfoque (sharpen)": np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]], dtype=float),
    "Sobel x": sobel_x,
    "Relieve (emboss)": np.array([[-2, -1, 0], [-1, 1, 1], [0, 1, 2]], dtype=float),
}

# Repetimos la comparación con tres productos para observar cómo responde cada filtro.
ejemplos = [("Antoni", ANTONI), ("KitKat", KITKAT), ("Zeus", ZEUS)]

for nombre_producto, imagen in ejemplos:
    # Convertimos a escala de grises y tomamos un recorte para validar la convolución manual.
    gris = cv.cvtColor(imagen, cv.COLOR_RGB2GRAY)
    alto, ancho = gris.shape
    recorte = gris[alto // 4:alto // 4 + min(100, alto // 2),
                   ancho // 4:ancho // 4 + min(100, ancho // 2)]

    # Comparamos la función manual con filter2D de OpenCV usando el mismo kernel.
    manual = convolucion(recorte, sobel_x)
    opencv = cv.filter2D(recorte.astype(float), -1, sobel_x, borderType=cv.BORDER_REPLICATE)
    print(f"{nombre_producto} - diferencia máxima convolución manual vs OpenCV:",
          np.abs(manual - opencv).max())

    # Aplicamos cada kernel a la imagen y mostramos el resultado junto con sus pesos.
    fig, ejes = plt.subplots(2, len(kernels), figsize=(18, 7.5))
    for c, (nombre, kernel) in enumerate(kernels.items()):
        salida = cv.filter2D(gris.astype(float), -1, kernel)

        # La magnitud absoluta permite visualizar los bordes detectados por Sobel.
        if nombre == "Sobel x":
            salida = np.abs(salida)

        ejes[0, c].imshow(np.clip(salida, 0, 255), cmap="gray", vmin=0, vmax=255)
        ejes[0, c].set_title(nombre)
        ejes[0, c].axis("off")

        # Representamos los valores del kernel para entender qué operación realiza.
        ejes[1, c].imshow(kernel, cmap="coolwarm", vmin=-2, vmax=2)
        if kernel.shape[0] == 3:
            for (i, j), valor in np.ndenumerate(kernel):
                ejes[1, c].text(j, i, f"{valor:g}", ha="center", va="center")
        ejes[1, c].set_title("Kernel")
        ejes[1, c].set_xticks([])
        ejes[1, c].set_yticks([])

    plt.suptitle(f"Convolución y filtros - {nombre_producto}")
    plt.tight_layout()
    plt.show()
```

### 2.2 a 2.5 Filtros de media, mediana y Gaussiano para reducir ruido

```python
# Comparamos la reducción de ruido en tres imágenes del dataset.
ejemplos = [("Antoni", ANTONI), ("KitKat", KITKAT), ("Zeus", ZEUS)]

# Definimos los filtros de media, mediana y Gaussiano.
filtros = {
    "Media 5x5": lambda im: cv.blur(im, (5, 5)),
    "Mediana 5x5": lambda im: cv.medianBlur(im, 5),
    "Gaussiano 5x5 σ=1": lambda im: cv.GaussianBlur(im, (5, 5), 1.0),
}

tabla = []
rng = np.random.default_rng(0)

for nombre_producto, imagen in ejemplos:
    # Trabajamos en escala de grises y generamos dos tipos de ruido artificial.
    limpia = cv.cvtColor(imagen, cv.COLOR_RGB2GRAY)
    ruido_gauss = np.clip(limpia.astype(float) + rng.normal(0, 25, limpia.shape), 0, 255).astype(np.uint8)

    ruido_sp = limpia.copy()
    mascara = rng.random(limpia.shape)
    ruido_sp[mascara < 0.04] = 0      # Pimienta: píxeles negros.
    ruido_sp[mascara > 0.96] = 255    # Sal: píxeles blancos.

    # Calculamos PSNR para comparar cuánto se parece cada resultado a la imagen limpia.
    for nombre_ruido, ruidosa in [
        ("Gaussiano (σ=25)", ruido_gauss),
        ("Sal y pimienta (8 %)", ruido_sp)
    ]:
        fila = {"producto": nombre_producto, "ruido": nombre_ruido,
                "sin filtrar": cv.PSNR(limpia, ruidosa)}
        for nombre_filtro, funcion in filtros.items():
            fila[nombre_filtro] = cv.PSNR(limpia, funcion(ruidosa))
        tabla.append(fila)

        # Mostramos la imagen con ruido y el resultado de cada filtro.
        mostrar(
            [ruidosa] + [funcion(ruidosa) for funcion in filtros.values()],
            [f"{nombre_producto}: {nombre_ruido}"] + list(filtros),
            None, 3.5, 1, 4
        )

# La tabla permite identificar qué filtro funciona mejor según el tipo de ruido.
print("PSNR en dB (un valor mayor indica mayor similitud con la imagen limpia):")
pd.DataFrame(tabla).set_index(["producto", "ruido"]).round(1)
```

## 3. Operaciones morfológicas

### 3.1 Aplicación: aislar un producto con máscara de color y morfología

```python
# Aplicamos la misma segmentación por color y las operaciones morfológicas a tres productos.
ejemplos = [("Antoni", ANTONI), ("KitKat", KITKAT), ("Zeus", ZEUS)]

for nombre_producto, imagen in ejemplos:
    # Convertimos la imagen a HSV para trabajar con el tono, la saturación y el brillo.
    hsv = cv.cvtColor(imagen, cv.COLOR_RGB2HSV)

    # Creamos una máscara que selecciona el rango de color definido para el producto.
    if nombre_producto == "Antoni":
        mascara = cv.bitwise_or(
            cv.inRange(hsv, (0, 80, 3), (15, 255, 100)),
            cv.inRange(hsv, (165, 80, 3), (179, 255, 100))
        )

    elif nombre_producto == "KitKat":
        mascara = cv.bitwise_or(
            cv.inRange(hsv, (0, 80, 40), (10, 255, 255)),
            cv.inRange(hsv, (170, 80, 40), (179, 255, 255))
        )

    elif nombre_producto == "Zeus":
        mascara = cv.inRange(hsv, (15, 15, 120), (35, 130, 255))

    # La apertura elimina regiones pequeñas y el cierre ayuda a rellenar huecos.
    elemento = cv.getStructuringElement(cv.MORPH_ELLIPSE, (7, 7))
    abierta = cv.morphologyEx(mascara, cv.MORPH_OPEN, elemento)
    cerrada = cv.morphologyEx(
        abierta, cv.MORPH_CLOSE,
        cv.getStructuringElement(cv.MORPH_ELLIPSE, (25, 25))
    )

    # Buscamos el contorno externo de mayor área y lo rellenamos para obtener la región principal.
    contornos, _ = cv.findContours(cerrada, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)
    producto = np.zeros_like(cerrada)
    caja = imagen.copy()

    if contornos:
        mayor = max(contornos, key=cv.contourArea)
        cv.drawContours(producto, [mayor], -1, 255, thickness=cv.FILLED)

        # Dibujamos una caja delimitadora alrededor del contorno seleccionado.
        x, y, w, h = cv.boundingRect(mayor)
        cv.rectangle(caja, (x, y), (x + w, y + h), (0, 255, 0), 3)

    # Mostramos cada etapa de la segmentación para comparar los tres productos.
    mostrar(
        [mascara, abierta, cerrada, producto, caja],
        ["1. Máscara HSV", "2. Apertura 7x7", "3. Cierre 25x25",
         "4. Contorno mayor", "5. Caja delimitadora"],
        None, 3, 1, 5
    )

    # Contamos cuántas regiones quedan después de cada operación.
    for titulo, resultado in [
        ("Máscara HSV", mascara), ("Tras apertura", abierta), ("Tras cierre", cerrada)
    ]:
        print(f"{nombre_producto} - {titulo}: {cv.connectedComponents(resultado)[0] - 1} regiones")
```

## Conclusión de la Unidad 2

El trabajo con el dataset de productos de supermercado permitió observar cómo las operaciones de procesamiento digital modifican y mejoran la información visual de las imágenes. Al aplicar las técnicas a tres productos diferentes —Antoni, KitKat y Zeus— se pudo comparar su comportamiento ante variaciones de color, iluminación, contraste y textura.

La ecualización global y CLAHE permiten mejorar el contraste, aunque CLAHE ofrece un ajuste local que ayuda a conservar mejor los detalles. La comparación de umbrales fijo, Otsu y adaptativo muestra que la binarización depende de las condiciones de iluminación de cada fotografía. Por otra parte, los kernels de convolución generan efectos distintos: algunos suavizan, otros realzan detalles y otros resaltan bordes. Los filtros de media, mediana y Gaussiano reducen el ruido, pero su eficacia depende del tipo de ruido presente.

Finalmente, las operaciones morfológicas permiten limpiar las máscaras binarias y destacar regiones de interés. En conjunto, estas técnicas constituyen una etapa importante de preprocesamiento, porque ayudan a obtener imágenes más uniformes y con información visual más útil para tareas posteriores de extracción de características y clasificación automática de productos.

Los resultados también muestran que no existe una única operación que funcione de la misma manera para todas las imágenes: es necesario observar cada producto y seleccionar los parámetros adecuados según sus características.
